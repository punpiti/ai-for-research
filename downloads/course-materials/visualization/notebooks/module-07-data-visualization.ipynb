## เริ่มแบบ standalone

ดาวน์โหลด notebook ทั้งชุดพร้อม `module-07-environment.yml` และ `module-07-requirements.txt`
หากใช้เครื่องของตนเอง ให้สร้าง environment และเลือก kernel ตาม
[คู่มือติดตั้ง Module 7](module-07-standalone-setup.md) หากใช้ Google Colab ให้อัปโหลด notebook
แล้วเลือก **Runtime → Run all** เซลล์ setup ของแต่ละ Part จะตรวจและติดตั้ง core package ให้ครบเอง

แต่ละ Part รันได้ด้วยตัวเอง ไม่ต้องเปิดหรือรัน Part ก่อนหน้า และไม่อาศัย package จากบทอื่น
